```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics.pairwise import cosine_similarity
print("Libraries imported successfully!")
```

```python
documents = [
    "Machine learning applications in computer vision and image processing.",
    "Neural network architectures for natural language processing tasks.",
    "Operating systems and memory management optimization techniques.",
    "Distributed systems and network protocols for multiplayer games.",
    "Supervised learning algorithms for predictive data analytics.",
    "Unsupervised clustering models for high-dimensional text datasets."
]

for i, doc in enumerate(documents):
    print(f"Doc {i+1}: {doc}")
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
X_vectors = vectorizer.fit_transform(documents)

print(f"Matrix shape (Documents x Features): {X_vectors.shape}")
```

```python
true_k = 2
model = KMeans(n_clusters=true_k, init='k-means++', random_state=42, n_init=10)
model.fit(X_vectors)

for doc_idx, label in enumerate(model.labels_):
    print(f"Document {doc_idx+1} belongs to Cluster: {label}")
```

```python
query = ["Deep learning models for text classification and NLP"]
query_vector = vectorizer.transform(query)
similarities = cosine_similarity(query_vector, X_vectors).flatten()

print(f"Query: '{query[0]}'\n")
for doc_idx, score in enumerate(similarities):
    print(f"Similarity to Doc {doc_idx+1} ({score:.4f}): {documents[doc_idx]}")

best_match_idx = np.argmax(similarities)
print(f"\n-> Best Match: Document {best_match_idx + 1} with a score of {similarities[best_match_idx]:.4f}")
```
